```python
import numpy as np
```

```python
# sample features
X = np.array([
    [2, 60, 45],
    [3, 65, 50],
    [4, 70, 55],
    [5, 75, 60],
    [6, 80, 65],
    [7, 85, 70],
    [8, 90, 80],
    [9, 95, 90]
], dtype=float)
```

```python
# target
y = np.array([0, 0, 0, 0, 1, 1, 1, 1], dtype=float).reshape(-1, 1)
```

```python
X = (X - X.mean(axis=0)) / X.std(axis=0)
```

```python
def relu(x):
    return np.maximum(0, x)
```

```python

def relu_derivative(x):
    return (x > 0).astype(float)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def binary_cross_entropy(y_true, y_pred):
    epsilon = 1e-7
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    return -np.mean(
        y_true * np.log(y_pred) +
        (1 - y_true) * np.log(1 - y_pred)
    )
```

```python
class DenseLayer:

    def __init__(self, input_size, output_size):
        self.weights = np.random.randn(input_size, output_size) * 0.01
        self.biases = np.zeros((1, output_size))

    def forward(self, X):
        self.input = X
        self.z = np.dot(X, self.weights) + self.biases
        return self.z
```

```python
class NeuralNetwork:

    def __init__(self):
        self.layer1 = DenseLayer(3, 4)
        self.layer2 = DenseLayer(4, 1)

    def forward(self, X):
        self.z1 = self.layer1.forward(X)
        self.a1 = relu(self.z1)

        self.z2 = self.layer2.forward(self.a1)
        self.output = sigmoid(self.z2)

        return self.output

    def backward(self, X, y, learning_rate):
        m = len(X)

        dz2 = self.output - y
        dw2 = np.dot(self.a1.T, dz2) / m
        db2 = np.sum(dz2, axis=0, keepdims=True) / m

        da1 = np.dot(dz2, self.layer2.weights.T)
        dz1 = da1 * relu_derivative(self.z1)

        dw1 = np.dot(X.T, dz1) / m
        db1 = np.sum(dz1, axis=0, keepdims=True) / m

        self.layer2.weights -= learning_rate * dw2
        self.layer2.biases -= learning_rate * db2

        self.layer1.weights -= learning_rate * dw1
        self.layer1.biases -= learning_rate * db1
```

```python
model = NeuralNetwork()
```

```python
epochs = 5000
learning_rate = 0.01
```

```python
for epoch in range(epochs):
    predictions = model.forward(X)
    loss = binary_cross_entropy(y, predictions)
    model.backward(X, y, learning_rate)

    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")
```

```python
predictions = model.forward(X)
```

```python
predicted_classes = (predictions >= 0.5).astype(int)
```

```python
print("\nPredicted:")
print(predicted_classes)
```

```python
print("\nActual:")
print(y.astype(int))
```

```python
accuracy = np.mean(predicted_classes == y) * 100
```

```python
print(f"\nAccuracy: {accuracy:.2f}%")
```

# Student Pass Prediction

Built a simple **Neural Network from scratch using NumPy** to predict student pass/fail outcomes based on study hours, attendance, and previous score.

### Process

**Input → Dense Layer → ReLU → Dense Layer → Sigmoid → Loss → Backpropagation → Gradient Descent → Prediction**

Used **Binary Cross-Entropy**, learning rate, epochs, weights, and biases to train the model.
